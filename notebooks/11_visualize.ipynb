# 11 · Visualise Population

Create publication-ready plots using the built-in plotting module.

Requires: `pip install matplotlib` (optional: `seaborn`)

```python
import logging
logging.basicConfig(level=logging.ERROR)

%matplotlib inline

from gbgsynth import GbgSynth, plotting

city = GbgSynth(year=2023)
haga = city.synthesize("Haga")
plotting.set_style("seaborn")
```

### Age distribution

```python
plotting.plot_age_distribution(haga, show_marginals=True)
```

### Household size

```python
plotting.plot_household_size(haga, show_marginals=True)
```

### Population pyramid

```python
plotting.plot_population_pyramid(haga)
```

### Car ownership

```python
plotting.plot_car_ownership(haga, by="household_size")
```

### Marginal comparison

```python
plotting.plot_marginal_comparison(haga)
```

### Error analysis

```python
plotting.plot_error_analysis(haga)
```

### Scatter comparison

```python
plotting.plot_scatter_comparison(haga)
```

### Compare areas side-by-side

```python
areas = [haga] + [city.synthesize(n) for n in ["Annedal", "Olivedal"]]

plotting.compare_areas(areas, metric="population")
```

```python
plotting.compare_areas(areas, metric="cars_per_capita")
```
