# 04 · Save Population to CSV

Generate a population and save individuals, households, and dwelling CSVs.

```python
import logging
logging.basicConfig(level=logging.ERROR)

from pathlib import Path
from gbgsynth import GbgSynth

# Output goes to <project_root>/output/
OUTPUT_DIR = str(Path.cwd().parent / "output")
OUTPUT_DIR
```

```python
city = GbgSynth(year=2023)
haga = city.synthesize("Haga")

saved = haga.save(output_dir=OUTPUT_DIR)

print("Saved files:")
for kind, path in saved.items():
    print(f"  {kind}: {path}")
```
