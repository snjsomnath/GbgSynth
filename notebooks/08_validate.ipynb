# 08 · Full Validation

Run out-of-sample validation that compares the synthetic population against census tables *not* used during fitting, plus sanity checks.

```python
import logging
logging.basicConfig(level=logging.ERROR, format="%(message)s")

from gbgsynth import GbgSynth
from gbgsynth.validation import Validator

city = GbgSynth(year=2023)
haga = city.synthesize("Haga")
```

### Run all validations

```python
validator = Validator(haga, tolerance=0.15)
report = validator.run_all_validations()

print(report.summary())
```

### Derived metrics

```python
print("Derived metrics:")
for name, value in validator.compute_derived_metrics().items():
    if isinstance(value, float):
        print(f"  {name}: {value:.3f}")
    else:
        print(f"  {name}: {value}")
```
