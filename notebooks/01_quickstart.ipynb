# 01 · Quick Start

Verify your installation and generate your first synthetic population in a few lines of code.

```python
import logging
logging.basicConfig(level=logging.ERROR)

from gbgsynth import GbgSynth
```

```python
city = GbgSynth(year=2023)
haga = city.synthesize("Haga")

print(f"City:  {city}")
print(f"Area:  {haga}")
print(f"Pop:   {len(haga.individuals):,} individuals")
print(f"HH:    {len(haga.households):,} households")
```

```python
print("Age statistics:")
haga.individuals_df["age"].describe()
```
